Our original plan was to see if we could predict C02 emissions, however we found that about 70% of the records were missing this data. Therefore, we chose to predict Plant annual net generation (MWh), which only had about 3% of missing records. Majority of the demographic data was below 5%.

To move forward, our hypothesis is that demographics and socioeconomic factors predict where a plant is located and its severity of output. We need to do EDA to see if Plant annual net generation (MWh), Plant annual total nonrenewables net generation (MWh), or  Plant annual total renewables net generation (MWh) would be the most appropriate to predict for

```python
fuelcat_vs_co2 = egrid_joined[['Plant primary fuel category_x',"Plant annual CO2 emissions (tons)"]].dropna()
fuelcat_vs_co2['Plant annual CO2 emissions (tons)'] = \
    fuelcat_vs_co2['Plant annual CO2 emissions (tons)'].str.replace(',', '').astype(int)
fuelcat_vs_co2
```

```python
plt.figure(figsize=(11.5, 6))
plt.xlabel("Plant primary fuel category")
ax = sns.boxplot(data= fuelcat_vs_co2, x= "Plant primary fuel category_x", \
                 y = "Plant annual CO2 emissions (tons)", log_scale=True)
ax;
```

```python
#clean the columns that are numerical (some contain , ( , ) )
relevant_columns = ['Plant primary fuel category_x', 'Plant annual net generation (MWh)', 'Total Population', 'People of Color (%)', 'Low Income (%)', 'Less Than High School Education (%)', 'Limited English Speaking (%)', 'Unemployment Rate (%)', 'Limited Life Expectancy (%)', 'Demographic Index', "National Percentile of Over Age 64"]
columns_to_clean = ["Plant annual CO2 emissions (tons)", 'Plant annual net generation (MWh)', 'Total Population', 'People of Color (%)', 'Low Income (%)', 'Less Than High School Education (%)', 'Limited English Speaking (%)', 'Unemployment Rate (%)', 'Limited Life Expectancy (%)', 'Demographic Index', "National Percentile of Over Age 64"]

egridf = egrid_joined[relevant_columns] #egrid final

for col in columns_to_clean:
    cleaned = egrid_joined[col].astype(str).str.replace(r'[,\(\)]', '', regex=True)
    egridf[col] = pd.to_numeric(cleaned, errors='coerce')
    egridf = egridf.dropna()
    egridf[col] = egridf[col].astype(int)
    

egridf.dropna()
egridf.shape
```

```python
plt.figure(figsize=(11.5, 6))
plt.xlabel("Plant primary fuel category")
ax = sns.boxplot(data= egridf, x= "Plant primary fuel category_x", \
                 y = "Plant annual net generation (MWh)", log_scale=True)
ax;
```

```python
gaussian_model_intercept = sm.GLM(
    egridf['Plant annual CO2 emissions (tons)'], sm.add_constant(egridf[num_cols]) ,
    family=sm.families.Gaussian()
    
)
gaussian_results = gaussian_model_intercept.fit()
print(gaussian_results.summary())
```

```python
gaussian_model_intercept_simple = sm.GLM(
    egridf['Plant annual CO2 emissions (tons)'], sm.add_constant(egridf[['People of Color (%)', 'Low Income (%)', 'Total Population']]) ,
    family=sm.families.Gaussian()
    
)
gaussian_results_simple = gaussian_model_intercept_simple.fit()
print(gaussian_results_simple.summary())
```
